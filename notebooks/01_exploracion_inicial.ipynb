```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().parent
file_path = project_root / "data" / "raw" / "retail_store_inventory.csv"

df = pd.read_csv(file_path)

df.head()

df["Date"] = pd.to_datetime(df["Date"])
df.info()
```

```python

print("tiendas:", df["Store ID"].nunique())
print("productos:", df["Product ID"].nunique())
print("categorías:", df["Category"].nunique())
print("regiones:", df["Region"].nunique())
print("climas:", df["Weather Condition"].nunique())
print("temporadas:", df["Seasonality"].nunique())
```

```python
print("\ncategorías:")
print(df["Category"].value_counts())

print("\nregiones:")
print(df["Region"].value_counts())

print("\nclimas:")
print(df["Weather Condition"].value_counts())

print("\ntemporadas:")
print(df["Seasonality"].value_counts())
```
